**Finding Missing Data**

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [4, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

**Removeing Missing data**

```python
df.dropna()
```

```python
df
```

```python
df.dropna(thresh=3)
```

**Filling the missing data**

```python
df.fillna(0)
```

```python
df                       #kintu main data thik ase so fillna(0,inplace=true) kore dile tokokn actual data teicng hye jbe
```

```python
values={'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```
